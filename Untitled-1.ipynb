```python
# # Task 2: Data Visualization with t-SNE
# Goal: visualize the preprocessed (high-dimensional) Rain in Australia data in 2D/3D
# using t-SNE, colored by the class label, and look for patterns or clusters.

# Imports and configuration 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.model_selection import train_test_split

X_PATH = "X_scaled_cleaned.csv"          # preprocessed features
Y_PATH = "y_cleaned.csv"          # labels (e.g. RainTomorrow)
SAMPLE_SIZE = 8000        # t-SNE is slow on ~140k rows, so subsample
PCA_COMPONENTS = 40       # PCA before t-SNE (denoise + speed)
PERPLEXITIES = [5, 30, 50]
RANDOM_STATE = 42
```

```python
# Load data 
X = pd.read_csv(X_PATH)
y = pd.read_csv(Y_PATH)

# y should be a single column -> Series
y = y.iloc[:, 0]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Class distribution:\n", y.value_counts(normalize=True).round(3)) # display class proportions
```

```python
# Additional checks
assert len(X) == len(y), "X and y have different numbers of rows!"
print("Missing values in X:", int(X.isna().sum().sum()))
print("Non-numeric columns in X:", list(X.select_dtypes(exclude=[np.number]).columns))

# Label names for the legend (handles both 0/1 and No/Yes labels)
classes = sorted(y.unique())
label_names = {c: str(c) for c in classes}
if set(classes) == {0, 1}:
    label_names = {0: "No rain (0)", 1: "Rain (1)"}
```

```python
# Stratified subsample 
# Keeps the class proportions the same as the full dataset.
if len(X) > SAMPLE_SIZE:
    X_s, _, y_s, _ = train_test_split(
        X, y, train_size=SAMPLE_SIZE, stratify=y, random_state=RANDOM_STATE
    )
else:
    X_s, y_s = X, y

sample_idx = X_s.index.values  # saved so Task 3 can reuse the same points
X_s_arr = X_s.values #converting to an array
y_s_arr = y_s.values
print("Subsample shape:", X_s_arr.shape)
```

```python
# (Optional) PCA before t-SNE 
if PCA_COMPONENTS is not None and X_s_arr.shape[1] > PCA_COMPONENTS:
    pca = PCA(n_components=PCA_COMPONENTS, random_state=RANDOM_STATE)
    X_in = pca.fit_transform(X_s_arr)
    print(f"PCA to {PCA_COMPONENTS} dims keeps "
          f"{pca.explained_variance_ratio_.sum():.1%} of the variance")
else:
    X_in = X_s_arr
    print("Skipping PCA; using all", X_in.shape[1], "features")
```

```python

# t-SNE (2D) at several perplexities 
embeddings = {}
fig, axes = plt.subplots(1, len(PERPLEXITIES), figsize=(6 * len(PERPLEXITIES), 5))
axes = np.atleast_1d(axes)

for ax, perp in zip(axes, PERPLEXITIES):
    tsne = TSNE(
        n_components=2,
        perplexity=perp,
        init="pca",
        learning_rate="auto",
        max_iter=1000,            
        random_state=RANDOM_STATE,
    )
    emb = tsne.fit_transform(X_in)
    embeddings[perp] = emb
    print(f"perplexity={perp}: KL divergence = {tsne.kl_divergence_:.3f}")

    for c in classes:
        m = y_s_arr == c
        ax.scatter(emb[m, 0], emb[m, 1], s=6, alpha=0.6, label=label_names[c])
    ax.set_title(f"t-SNE (perplexity = {perp})")
    ax.set_xlabel("t-SNE dimension 1")
    ax.set_ylabel("t-SNE dimension 2")
    ax.legend(title="RainTomorrow", markerscale=3)

plt.tight_layout()
plt.savefig("tsne_perplexity_comparison.png", dpi=200)
plt.show()
```

```python
# Main plot (pick the perplexity you like best) 
BEST_PERP = 30
emb = embeddings[BEST_PERP]

plt.figure(figsize=(8, 6))
for c in classes:
    m = y_s_arr == c
    plt.scatter(emb[m, 0], emb[m, 1], s=8, alpha=0.6, label=label_names[c])
plt.title(f"t-SNE projection of Rain in Australia (perplexity = {BEST_PERP})")
plt.xlabel("t-SNE dimension 1")
plt.ylabel("t-SNE dimension 2")
plt.legend(title="RainTomorrow", markerscale=3)
plt.tight_layout()
plt.savefig("tsne_main.png", dpi=200)
plt.show()
```

```python

# Color by individual features to interpret the structure
# Helps explain WHAT drives the clusters. Uses columns if they exist in X.
candidate_feats = ["Humidity3pm", "Pressure3pm", "Rainfall", "Temp3pm", "Sunshine"]
feats = [f for f in candidate_feats if f in X_s.columns]
if not feats:                                   # fall back to the first 4 columns
    feats = list(X_s.columns[:4])

fig, axes = plt.subplots(1, len(feats), figsize=(5 * len(feats), 4.5))
axes = np.atleast_1d(axes)
for ax, f in zip(axes, feats):
    sc = ax.scatter(emb[:, 0], emb[:, 1], c=X_s[f].values, s=6, alpha=0.7, cmap="viridis")
    ax.set_title(f"Colored by {f}")
    ax.set_xlabel("t-SNE 1")
    ax.set_ylabel("t-SNE 2")
    plt.colorbar(sc, ax=ax)
plt.tight_layout()
plt.savefig("tsne_by_feature.png", dpi=200)
plt.show()
```

```python

# (Optional) 3D t-SNE
RUN_3D = True
if RUN_3D:
    tsne3d = TSNE(n_components=3, perplexity=BEST_PERP, init="pca",
                  learning_rate="auto", random_state=RANDOM_STATE)
    emb3d = tsne3d.fit_transform(X_in)

    fig = plt.figure(figsize=(8, 7))
    ax = fig.add_subplot(111, projection="3d")
    for c in classes:
        m = y_s_arr == c
        ax.scatter(emb3d[m, 0], emb3d[m, 1], emb3d[m, 2], s=5, alpha=0.5,
                   label=label_names[c])
    ax.set_title("3D t-SNE projection")
    ax.set_xlabel("t-SNE 1"); ax.set_ylabel("t-SNE 2"); ax.set_zlabel("t-SNE 3")
    ax.legend(title="RainTomorrow", markerscale=3)
    plt.tight_layout()
    plt.savefig("tsne_3d.png", dpi=200)
    plt.show()
```

```python

# Save embedding + sample indices for Task 3 (clustering) 
out = pd.DataFrame(emb, columns=["tsne1", "tsne2"], index=sample_idx)
out["label"] = y_s_arr
out.to_csv("tsne_embedding_perp30.csv")
print("Saved embedding for", len(out), "points")
```
